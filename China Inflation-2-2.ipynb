# One Month Ahead Forecasting – Inflation Rate China

# 1: Preprocessing

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import TimeSeriesSplit
```

```python

df = pd.read_csv(
    "china_economic_raw.csv",
    sep=";",
    na_values=["#DIV/0!", "#N/A", "#VALUE!", "#REF!"]
)

df.columns = df.columns.str.strip()

# Convert Date column
df["Date"] = pd.to_datetime(df["Date"], format="%b %y")

# Sort chronologically
df = df.sort_values("Date").reset_index(drop=True)

# Define target and predictors
target = "INFL"
predictors = [
    col for col in df.columns
    if col.startswith("X") and col[1:].isdigit()
]

# Basic inspection
print(df.shape)
print(df.head())
print(df.info())
print(df.describe())
```

```python
print(df.columns.tolist())
```

In order to run the following code, convert Strings to numbers.

```python
# Check which predictors are not numeric
non_numeric = df[predictors].select_dtypes(
    exclude="number"
).columns

print("Non-numeric features:")
print(non_numeric.tolist())
```

```python
# Convert predictors and target to numeric values

for col in predictors + ["INFL"]:
    original = df[col]

    converted = pd.to_numeric(
        original.astype("string")
        .str.strip()
        .str.replace(",", ".", regex=False),
        errors="coerce"
    )

    # Check if conversion introduced additional NaNs
    new_missing = original.notna() & converted.isna()

    if new_missing.any():
        print(f"{col}: Unexpected values:")
        print(original[new_missing].unique()[:10])

    df[col] = converted

# Check whether all variables are numeric
print(df[predictors].dtypes.value_counts())
```

```python
missing_values = df[predictors].isna().sum()

missing_percentage = (
    df[predictors].isna().mean() * 100
)

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing %": missing_percentage
})

print(
    missing_report.sort_values(
        "Missing %", ascending=False
    ).head(20)
)
```

High  portion of missing values indicates the necessity of further analysing nature of data. Visual inspection of dataset hints to more missing values earlier.
If this is true, which will be tested, it could be beneficial to use only a later part of the dataset, since the data quality significantly improves and by this the model would probably also significantly improve (garbage in, garbage out).

```python
# Calculate percentage of missing values for each year

yearly_missing = (
    df[predictors]
    .isna()
    .groupby(df["Date"].dt.year)
    .mean()
    .mean(axis=1) * 100
)

# Display results
print(yearly_missing.round(2).to_string())
```

From 2004 to 2005, the amount of missing values decreased significantly. For this reason, we will only use data from 2005 onwards to train and test our model. We decided to use this threshold because the completeness of the data improved a lot by almost 10 percentage points. However, we want to point out that it might also have been possible to use the data from 2000 onwards (see report). For now, we have decided to retain observations prior to 2005 for the construction of lagged features, while excluding them as separate training observations. This allows us to preserve potentially useful historical information that would otherwise be lost. Although more recent observations may be more predictive, some economic indicators might also influence inflation with longer time lags. We will therefore evaluate the relevance of different lags during model training and validation.

```python
# Create a copy of our dataset
data = df.sort_values("Date").reset_index(drop=True).copy()

# Keep original values for duplicate detection later
original_values = data[predictors].copy()

# Missing values before forward fill
missing_before = data[predictors].isna().sum().sum()

# Forward fill, maximum 2 consecutive months
data[predictors] = data[predictors].ffill(limit=2)

# Missing values after forward fill
missing_after = data[predictors].isna().sum().sum()

print("Missing values before:", missing_before)
print("Missing values after:", missing_after)
```

We decided to use forward fill to replace short gaps of up to two consecutive months with the last available value. Since forward fill only uses previous observations, we do not use any future data to fill missing values. We limit it to two months because using older observations for longer gaps might be less reliable. Longer gaps will remain as NaN for now.

We also want to remove duplicate features since they do not provide additional information to our models. Before removing them, we check if the variables have identical values wherever both have observations. If they do, we first combine them so that we do not lose any data if one column has missing values that are available in the other column. We then remove one of the duplicate columns.

```python
# Create a fresh copy of the dataset
data = df.sort_values("Date").reset_index(drop=True).copy()

# Convert all predictors to numeric
for col in predictors:
    original = data[col]

    converted = pd.to_numeric(
        original.astype("string")
        .str.strip()
        .str.replace(",", ".", regex=False),
        errors="coerce"
    )

    # Check whether conversion creates new missing values
    new_missing = original.notna() & converted.isna()

    if new_missing.any():
        raise ValueError(
            f"Non-numeric values found in {col}: "
            f"{original[new_missing].unique()[:10]}"
        )

    data[col] = converted.astype(float)

# Keep original numeric values for duplicate detection
original_values = data[predictors].copy()

# Forward fill short gaps
data[predictors] = data[predictors].ffill(limit=2)

# Detect and merge duplicate features
features = predictors.copy()
merged_pairs = []

for i, keep in enumerate(predictors):
    if keep not in features:
        continue

    for remove in predictors[i + 1:]:
        if remove not in features:
            continue

        a = original_values[keep]
        b = original_values[remove]

        # Compare only observations where both have values
        overlap = a.notna() & b.notna()

        if overlap.sum() < 24:
            continue

        if a[overlap].nunique() < 3:
            continue

        # Check whether values match
        if np.allclose(
            a[overlap].to_numpy(dtype=float),
            b[overlap].to_numpy(dtype=float),
            rtol=1e-9,
            atol=1e-8
        ):

            # Combine available observations
            merged = a.combine_first(b)

            original_values[keep] = merged
            data[keep] = merged.ffill(limit=2)

            # Remove redundant column
            data = data.drop(columns=remove)
            features.remove(remove)

            merged_pairs.append((keep, remove))

print("Merged duplicate columns:", merged_pairs)
print("Remaining features:", len(features))
```

For the initial feature selection, we focus on missing values between 2005 and 2020 rather than over the entire dataset. Since we plan to use observations from 2005 onwards for our predictions, we do not want to remove features just because they had many missing values in the 1990s. These earlier values might still be useful for lags, but the completeness of the features in our main observation period is more relevant.

For now, we use a threshold of 20% missing values. We also remove constant features since they do not provide any information about changes in our target variable.

```python

# Reference period for initial feature filtering
reference_mask = data["Date"].between(
    "2005-01-01", "2023-12-01"
)

# Missing values per feature
missing_ratio = (
    data.loc[reference_mask, features]
    .isna()
    .mean()
)

# Display highest missing ratios
print(
    (missing_ratio * 100)
    .sort_values(ascending=False)
    .head(20)
)

# Remove features with more than 20% missing
threshold = 0.20

drop_missing = missing_ratio[
    missing_ratio > threshold
].index.tolist()

# Identify constant features
drop_constant = [
    col for col in features
    if data.loc[
        reference_mask, col
    ].nunique(dropna=True) <= 1
]

# Combine exclusions
drop_features = list(set(
    drop_missing + drop_constant
))

# Remove features
data = data.drop(columns=drop_features)

features = [
    col for col in features
    if col not in drop_features
]

print("Dropped for missing values:", drop_missing)
print("Dropped constant features:", drop_constant)
print("Remaining features:", len(features))
```

No variable was removed because of missing values for now. Since 6.25% is the highest portion of missing values in all of the given features we can keep all of them for a start.

Lastly, we want to check whether some features directly explain our target variable. If a feature contains the same information as INFL, using it could lead to data leakage and unrealistically good predictions. We therefore check whether any of our features have an exact linear relationship with INFL and remove them if necessary.

```python

target = "INFL"

leakage_features = []

for col in features:

    # Use observations where both values exist
    pair = data.loc[
        reference_mask, [col, target]
    ].dropna()

    if len(pair) < 24 or pair[col].nunique() <= 1:
        continue

    x = pair[col].to_numpy()
    y = pair[target].to_numpy()

    # Test whether INFL = a * X + b
    A = np.column_stack([x, np.ones_like(x)])

    slope, intercept = np.linalg.lstsq(
        A, y, rcond=None
    )[0]

    predicted = slope * x + intercept

    max_error = np.max(np.abs(y - predicted))

    if max_error < 1e-8:
        leakage_features.append(col)

print("Directly explaining features:", leakage_features)
```

None of the individual features were found to have an exact linear relationship with the target variable. However, this does not rule out other forms of data leakage. To avoid using information from the same month as our prediction target, we will only use lagged predictor variables.

For prediction, we use lagged features. We move all features by one month.

```python
# Shift all predictors by one month
X = data[features].shift(1).copy()

X.columns = [
    col + "_lag1" for col in features
]
```

We will add some potentially useful predictors to the already existing ones. For one, previous inflation rates might be a useful predictor of this month's inflation. We use last month's inflation and inflation from 3 and 12 months ago. We refrained from any other engineered features since, in relation to our number of observations, we already have a large number of features. Moreover, we plan on using XGBoost and decision trees, which can learn non-linear relationships from the input data.

```python
# Historical inflation
X["INFL_lag1"] = data["INFL"].shift(1)
X["INFL_lag3"] = data["INFL"].shift(3)
X["INFL_lag12"] = data["INFL"].shift(12)

# Average inflation over the previous 3 months
X["INFL_mean3"] = (
    data["INFL"].shift(1).rolling(3).mean()
)
```

Because seasonality could be a relevant factor in inflation, we decode the month of the observation.

```python
month = data["Date"].dt.month

X["month_sin"] = np.sin(2 * np.pi * month / 12)
X["month_cos"] = np.cos(2 * np.pi * month / 12)
```

```python
# Combine target and engineered features
prepared = pd.concat(
    [data[["Date", "INFL"]], X],
    axis=1
)

# Keep only observations from 2005 onward
prepared = prepared[
    (prepared["Date"] >= "2005-01-01") &
    prepared["INFL"].notna()
].copy()

# Training observations only
outlier_train = prepared[
    prepared["Date"] < "2024-01-01"
].copy()

# Exclude binary and seasonal features from IQR audit
outlier_cols = [
    col for col in X.columns
    if col not in [
        "PMI_expansion_lag1",
        "month_sin",
        "month_cos"
    ]
]

Q1 = outlier_train[outlier_cols].quantile(0.25)
Q3 = outlier_train[outlier_cols].quantile(0.75)
IQR = Q3 - Q1

outliers = (
    (outlier_train[outlier_cols] < Q1 - 1.5 * IQR) |
    (outlier_train[outlier_cols] > Q3 + 1.5 * IQR)
)

print(outliers.sum().sort_values(ascending=False).head(20))
```

We used the IQR method to identify potential outliers in our predictor variables. However, we decided not to remove them for now. Since we are working with economic data, extreme values might reflect actual economic events rather than errors in the dataset. Removing these observations could therefore lead to a loss of relevant information for predicting inflation. If necessary, we might reconsider how to handle outliers depending on the performance of our models.

Up to this point, we have not replaced all missing values.

```python
# Analyze remaining missing values after feature engineering

missing_values = prepared[X.columns].isna().sum()

missing_percentage = (
    prepared[X.columns].isna().mean() * 100
)

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing %": missing_percentage
})

print(
    missing_report[
        missing_report["Missing Values"] > 0
    ].sort_values(
        "Missing %", ascending=False
    )
)

print(
    "Total remaining missing values:",
    missing_values.sum()
)
```

We will handle these missing values with methods like median imputation during the training of the models to avoid data leakage.

The next step is, do define a uniform test and train split. After that, the models can be trained and afterwards tested on that data. Further data processing techniques that are fitted to the different models will be used in the respective pipelines.

```python
# Chronological train-test split

# Training: January 2005 - December 2023
train = prepared[
    (prepared["Date"] >= "2005-01-01") &
    (prepared["Date"] < "2024-01-01")
].copy()

# Testing: January 2024 - April 2026
test = prepared[
    (prepared["Date"] >= "2024-01-01") &
    (prepared["Date"] <= "2026-04-01")
].copy()

# Separate predictors and target
X_train = train.drop(columns=["Date", "INFL"])
y_train = train["INFL"]

X_test = test.drop(columns=["Date", "INFL"])
y_test = test["INFL"]

# Check number of observations
print("Training observations:", len(train))
print("Testing observations:", len(test))
```

We decided to use the time period from 2005 to 2023 for training purposes, which contains a total of 228 observations. For testing, we use the data from January 2024 to April 2026 (28 data points).

We decided not to use a separate validation set, as we will use time-series cross-validation for hyperparameter tuning within our training data. This also allows us to use more observations for training, which might be beneficial given the relatively large number of predictor variables.

We chose this split because we want to keep as many observations as possible for training while still having a sufficiently long period of unseen data to evaluate our models. The test period covers 28 consecutive months, allowing us to evaluate how well our models predict inflation on more recent data.

Since we are working with time-series data, we also need to make sure that the training data always comes before the testing data to avoid data leakage.

# Decision Tree Regression

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import GridSearchCV, TimeSeriesSplit

dt_pipeline = Pipeline([("imputer", SimpleImputer(strategy="median")), ("tree", DecisionTreeRegressor(random_state=42))])

# Candidate levels of tree complexity
dt_param_grid = { "tree__max_depth": [2, 3, 4, 5, None], "tree__min_samples_leaf": [5, 10, 20]}

# Preserve chronological order during validation
dt_cv = TimeSeriesSplit(n_splits=5)
dt_search = GridSearchCV(estimator=dt_pipeline,param_grid=dt_param_grid,scoring="neg_root_mean_squared_error",cv=dt_cv, n_jobs=-1, refit=True)
dt_search.fit(X_train, y_train)

# refits the selected pipeline on all training observations
best_dt = dt_search.best_estimator_

print("Best parameters:", dt_search.best_params_)
print("Mean validation RMSE:", -dt_search.best_score_)
```

```python
from sklearn.metrics import (mean_squared_error,mean_absolute_error,r2_score)

dt_train_pred = best_dt.predict(X_train)
dt_test_pred = best_dt.predict(X_test)

# Persistence benchmark: predict using last month's inflation
naive_test_pred = X_test["INFL_lag1"].to_numpy(dtype=float)

def regression_metrics(actual, predicted):
    return { "RMSE": np.sqrt(mean_squared_error(actual, predicted)), "MAE": mean_absolute_error(actual, predicted), "R2": r2_score(actual, predicted)}

dt_results = pd.DataFrame({ "Decision Tree — Train": regression_metrics( y_train, dt_train_pred ),"Decision Tree — Test": regression_metrics(y_test, dt_test_pred),"Last-month benchmark — Test": regression_metrics(y_test, naive_test_pred)}).T

display(dt_results.round(4))
```

```python
plt.figure(figsize=(12, 5))

plt.plot(test["Date"], y_test, label="Actual inflation", marker="o")
plt.plot(test["Date"], dt_test_pred,label="Decision Tree forecast", marker="o")

plt.title("China: One-Month-Ahead Inflation Forecast")
plt.xlabel("Target month")
plt.ylabel("Monthly inflation (%)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
from sklearn.tree import plot_tree

# Extract the trained tree and its input feature names
dt_tree = best_dt.named_steps["tree"]

dt_feature_names = (best_dt.named_steps["imputer"].get_feature_names_out(X_train.columns))

plt.figure(figsize=(16, 7))

plot_tree(dt_tree,feature_names=list(dt_feature_names),filled=True,rounded=True,impurity=False,precision=3,fontsize=10)

plt.title("Decision Tree: Rules for Predicting Monthly Inflation")
plt.tight_layout()
plt.show()
```

```python
benchmark_predictions = {"Decision Tree": dt_test_pred,

    # Predict last month's inflation
    "Last-month inflation":
        X_test["INFL_lag1"].to_numpy(dtype=float),

    # Always predict the mean calculated on training data
    "Training mean":
        np.full(len(y_test), float(y_train.mean())),

    # Predict inflation observed in the same month one year earlier
    "Same month last year":
        X_test["INFL_lag12"].to_numpy(dtype=float)
}

dt_comparison = pd.DataFrame({name: regression_metrics(y_test, predictions)for name, predictions in benchmark_predictions.items()}).T

display(dt_comparison.round(4))
```

```python
# Explore both simpler and more complex trees
dt_diagnostic_grid = {"tree__max_depth": [1, 2, 3, 4, 5, 8, None],"tree__min_samples_leaf": [1, 2, 5, 10, 20, 30]}

dt_diagnostic_search = GridSearchCV(estimator=dt_pipeline,param_grid=dt_diagnostic_grid,scoring="neg_root_mean_squared_error",cv=dt_cv,return_train_score=True,n_jobs=-1,refit=False)

dt_diagnostic_search.fit(X_train, y_train)
```

```python
cv_results = pd.DataFrame(dt_diagnostic_search.cv_results_)

dt_diagnostics = pd.DataFrame({"max_depth": cv_results["param_tree__max_depth"],"min_samples_leaf": cv_results["param_tree__min_samples_leaf"],"Train RMSE": -cv_results["mean_train_score"],"Validation RMSE": -cv_results["mean_test_score"]})

dt_diagnostics["Gap"] = (dt_diagnostics["Validation RMSE"]- dt_diagnostics["Train RMSE"])

display(dt_diagnostics.sort_values("Validation RMSE").head(10).round(4))
```

The decision tree achieved a test RMSE of 0.363 percentage points, outperforming the historical-mean, previous-month and same-month-last-year benchmarks. It captured several large inflation movements but underestimated their magnitude, while its four terminal leaves produced relatively flat forecasts in other months. Increasing tree complexity did not improve cross-validation performance, supporting the simpler specification. Predictions relied on inflation lagged by twelve months and indicator X160 lagged by one month; these relationships are predictive rather than causal. Overall, the results suggest useful forecasting performance over the 28-month test period, although broader reliability remains uncertain.

# Ridge Regression and Elastic Net

Ridge regression and Elastic Net are linear models that penalise large coefficients. This matters here because there are 167 predictors but only 228 training observations, and many of the predictors are highly correlated, so a plain linear regression would overfit. Ridge shrinks all coefficients towards zero but keeps every predictor. Elastic Net mixes the Ridge and Lasso penalties and can also set coefficients to exactly zero, which removes predictors from the model. The parameter l1_ratio controls the mix, where 1.0 is a pure Lasso.

Both models are sensitive to the scale of the features, so the pipelines first fill missing values with the median and then standardise. Both steps are fitted inside each cross-validation fold to avoid data leakage. The hyperparameters are tuned with the same TimeSeriesSplit as for the decision tree, and the train/test split stays the same.

```python
from sklearn.linear_model import Ridge, ElasticNet
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

ridge_cv = TimeSeriesSplit(n_splits=5)

# Ridge: imputation -> standardisation -> L2-penalised regression
ridge_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("ridge", Ridge(max_iter=10000))
])

# Wide logarithmic grid for the penalty strength
ridge_param_grid = {"ridge__alpha": np.logspace(-2, 6, 50)}

ridge_search = GridSearchCV(
    estimator=ridge_pipeline,
    param_grid=ridge_param_grid,
    scoring="neg_root_mean_squared_error",
    cv=ridge_cv,
    return_train_score=True,
    n_jobs=-1,
    refit=True
)
ridge_search.fit(X_train, y_train)

best_ridge = ridge_search.best_estimator_

print("Best parameters:", ridge_search.best_params_)
print("Mean validation RMSE:", -ridge_search.best_score_)
```

```python
# Validation RMSE as a function of the Ridge penalty
ridge_cv_results = pd.DataFrame(ridge_search.cv_results_)

plt.figure(figsize=(10, 5))
plt.semilogx(ridge_cv_results["param_ridge__alpha"].astype(float),
             -ridge_cv_results["mean_train_score"], label="Training RMSE (CV folds)")
plt.semilogx(ridge_cv_results["param_ridge__alpha"].astype(float),
             -ridge_cv_results["mean_test_score"], label="Validation RMSE (CV folds)")
plt.axvline(ridge_search.best_params_["ridge__alpha"], color="grey", linestyle="--", label="Selected alpha")
plt.title("Ridge: Effect of the Penalty Strength")
plt.xlabel("alpha (log scale)")
plt.ylabel("RMSE (percentage points)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Elastic Net: imputation -> standardisation -> L1/L2-penalised regression
en_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("en", ElasticNet(max_iter=50000, random_state=42))
])

# Penalty strength and L1/L2 mixture
en_param_grid = {
    "en__alpha": np.logspace(-3, 1, 30),
    "en__l1_ratio": [0.1, 0.3, 0.5, 0.7, 0.9, 1.0]
}

en_search = GridSearchCV(
    estimator=en_pipeline,
    param_grid=en_param_grid,
    scoring="neg_root_mean_squared_error",
    cv=ridge_cv,
    return_train_score=True,
    n_jobs=-1,
    refit=True
)
en_search.fit(X_train, y_train)

best_en = en_search.best_estimator_

print("Best parameters:", en_search.best_params_)
print("Mean validation RMSE:", -en_search.best_score_)
```

```python
# Cross-validation results of the Elastic Net grid (best combinations)
en_cv_results = pd.DataFrame(en_search.cv_results_)

en_top = pd.DataFrame({
    "alpha": en_cv_results["param_en__alpha"].astype(float),
    "l1_ratio": en_cv_results["param_en__l1_ratio"].astype(float),
    "Train RMSE": -en_cv_results["mean_train_score"],
    "Validation RMSE": -en_cv_results["mean_test_score"]
}).sort_values("Validation RMSE").head(10)

display(en_top.round(4))
```

```python
# Test-set evaluation against the same benchmarks used for the decision tree
ridge_train_pred = best_ridge.predict(X_train)
ridge_test_pred = best_ridge.predict(X_test)
en_train_pred = best_en.predict(X_train)
en_test_pred = best_en.predict(X_test)

linear_train_results = pd.DataFrame({
    "Ridge: Train": regression_metrics(y_train, ridge_train_pred),
    "Elastic Net: Train": regression_metrics(y_train, en_train_pred)
}).T
display(linear_train_results.round(4))

linear_predictions = {
    "Ridge": ridge_test_pred,
    "Elastic Net": en_test_pred,
    "Decision Tree": dt_test_pred,
    "Last-month inflation": X_test["INFL_lag1"].to_numpy(dtype=float),
    "Training mean": np.full(len(y_test), float(y_train.mean())),
    "Same month last year": X_test["INFL_lag12"].to_numpy(dtype=float)
}

linear_comparison = pd.DataFrame({
    name: regression_metrics(y_test, pred)
    for name, pred in linear_predictions.items()
}).T

display(linear_comparison.round(4))
```

```python
plt.figure(figsize=(12, 5))

plt.plot(test["Date"], y_test, label="Actual inflation", marker="o", color="black")
plt.plot(test["Date"], ridge_test_pred, label="Ridge forecast", marker="o")
plt.plot(test["Date"], en_test_pred, label="Elastic Net forecast", marker="o")

plt.title("China: One-Month-Ahead Inflation Forecast (Ridge and Elastic Net)")
plt.xlabel("Target month")
plt.ylabel("Monthly inflation (%)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Coefficients on standardised predictors
ridge_coef = pd.Series(best_ridge.named_steps["ridge"].coef_, index=X_train.columns)
en_coef = pd.Series(best_en.named_steps["en"].coef_, index=X_train.columns)

print("Ridge: non-zero coefficients:", int((ridge_coef != 0).sum()), "of", len(ridge_coef))
print("Elastic Net: non-zero coefficients:", int((en_coef != 0).sum()), "of", len(en_coef))

print("\nElastic Net: selected predictors")
display(en_coef[en_coef != 0].sort_values(key=np.abs, ascending=False).to_frame("Coefficient").round(4))

top_ridge = ridge_coef.reindex(ridge_coef.abs().sort_values(ascending=False).index).head(10)

plt.figure(figsize=(9, 5))
top_ridge.sort_values().plot(kind="barh")
plt.title("Ridge: Ten Largest Standardised Coefficients")
plt.xlabel("Coefficient")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

Ridge selected the largest penalty in the grid (alpha = 1,000,000). With this penalty all coefficients are close to zero, so the model just predicts the training mean. The test results are therefore the same as for the training-mean benchmark (RMSE 0.517, R² -0.016). Weaker penalties fitted the training data better but performed worse on the later validation folds, so Ridge does not find useful signal among the 167 correlated predictors. A larger grid would not change this, since the model already equals the mean forecast.

Elastic Net did better. It selected alpha = 0.161 and l1_ratio = 0.9, which is close to a Lasso, and kept only two of the 167 predictors: inflation lagged by twelve months (negative coefficient) and X161 lagged by one month (very small coefficient). The test RMSE is 0.409 percentage points with an R² of 0.366, which beats Ridge and all three benchmarks. The decision tree is still better (RMSE 0.363, R² 0.499), so non-linear effects seem to matter somewhat. All models rely mainly on inflation lagged by twelve months. The tree uses X160 and Elastic Net uses X161, which are different indicators, and the very small X161 coefficient should not be interpreted economically.

The test period has only 28 observations, so small differences in RMSE between the models should not be over-interpreted. The Elastic Net R² is higher on the test set (0.366) than on the training set (0.235), which suggests that the test months are easier to forecast. As in the other sections, the results describe predictive relationships, not causal effects.
